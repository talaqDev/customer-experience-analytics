# Sentiment Analysis of Ethiopian Mobile Banking Apps
**Apps analyzed:**
- Commercial Bank of Ethiopia (CBE)
- Bank of Abyssinia (BOA)
- Dashen Bank (Amole)

**Date:** June 09, 2025

This notebook performs basic sentiment analysis on Play Store user reviews for the three mobile banking apps. It uses TextBlob and basic NLP techniques to extract and visualize sentiment.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from textblob import TextBlob
from wordcloud import WordCloud

# For inline plotting
%matplotlib inline
```

## Load Review Data

```
# Placeholder: Load your CSV files with reviews
# Each CSV should have at least: 'app_name', 'review', 'rating'

# Example structure:
# df = pd.read_csv('ethiopian_banks_reviews.csv')

# For demonstration, create sample data
data = {
    'app_name': ['CBE Birr', 'BOA', 'Amole', 'CBE Birr', 'BOA', 'Amole'],
    'review': [
        "Great app, very useful and easy to use.",
        "Not working properly, always crashing.",
        "Amazing features for paying bills.",
        "Login fails every time. Very frustrating.",
        "Smooth transactions and great UI.",
        "Terrible network error all the time."
    ]
}
df = pd.DataFrame(data)
df.head()
```

## Perform Sentiment Analysis

```
def get_sentiment(text):
    analysis = TextBlob(text)
    return analysis.sentiment.polarity

df['sentiment_score'] = df['review'].apply(get_sentiment)

# Label as Positive, Neutral, Negative
def get_sentiment_label(score):
    if score > 0.1:
        return 'Positive'
    elif score < -0.1:
        return 'Negative'
    else:
        return 'Neutral'

df['sentiment'] = df['sentiment_score'].apply(get_sentiment_label)
df.head()
```

## Sentiment Distribution by App

```
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='app_name', hue='sentiment', palette='Set2')
plt.title('Sentiment Distribution by App')
plt.xlabel('App Name')
plt.ylabel('Review Count')
plt.legend(title='Sentiment')
plt.tight_layout()
plt.show()
```

## Word Cloud of Reviews

```
text = ' '.join(df['review'])

wordcloud = WordCloud(width=800, height=400, background_color='white').generate(text)

plt.figure(figsize=(10, 5))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.title('Word Cloud of All Reviews')
plt.show()
```
